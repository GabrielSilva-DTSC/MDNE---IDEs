# Análise de sentimentos em comentários do YouTube

Projeto de modelagem de dados não estruturados. O coletor usa a YouTube Data API v3 para analisar o sentimento geral dos comentários de cada vídeo.

## Etapas do projeto

1. Definir a pergunta de pesquisa, os vídeos e o recorte de comentários.
2. Executar `python dados/brutos/bronze.py` na raiz para coletar até 500 comentários e respostas por vídeo em `dados/brutos/youtube/`.
3. Preparar os textos com os módulos de `src/preprocessamento/`.
4. Salvar dados preparados em `dados/processados/`.
5. Classificar sentimentos com `src/modelagem/sentimentos.py`.
6. Analisar resultados com `src/analise/estatisticas.py`.
7. Registrar decisões em `diario_de_bordo/` e resultados em `relatorio/`.

O coletor está implementado usando apenas a biblioteca padrão do Python. Os demais módulos ainda são esboços. Antes da coleta, configure `YOUTUBE_API_KEY` no arquivo `.env` da raiz do projeto. Consulte o README para entender os parâmetros, a paginação, os arquivos de saída e os testes.
